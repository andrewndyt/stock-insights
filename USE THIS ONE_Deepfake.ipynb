# DeepFake

Deepfake is a technology that uses artificial intelligence to manipulate the appearance and voice of a person in a video. Roop uses a face swapping technique that replaces the original face in the video with the desired face, while preserving the facial expressions and movements.

# Roop Repo
Roop is a repository on GitHub that allows users to create deepfake videos using only one image of the desired face.

Roop was developed by s0md3v, a security researcher and developer from India1. He started the project in May 2021 and discontinued it in August 2023, citing ethical issues and breach of trust by another developer who published a problematic video to the documentation of the project1. Roop has received over 400 commits and 2,000 stars on GitHub1. It has also been forked by other developers who have made some modifications and improvements2

````
%%writefile /content/roop/run_swap.py
import os
import sys
import cv2
import numpy as np
import subprocess
import onnxruntime
import insightface
from insightface.app import FaceAnalysis

def merge_audio(swapped_video, original_video):
    temp_output = swapped_video.replace('.mp4', '_with_audio.mp4')
    print("Merging original audio into the face-swapped video...")

    cmd = [
        'ffmpeg', '-y',
        '-i', swapped_video,
        '-i', original_video,
        '-c:v', 'copy',
        '-c:a', 'aac',
        '-map', '0:v:0',
        '-map', '1:a:0',
        '-shortest',
        temp_output
    ]

    result = subprocess.run(cmd, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
    if result.returncode == 0:
        os.replace(temp_output, swapped_video)
        print("Successfully merged original audio into the output!")
    else:
        print("Audio merge failed, keeping silent output. Error detail:")
        print(result.stderr)

def main():
    source_path = '/content/contentimage.jpeg.jpg'
    target_path = '/content/contentvideo_40.mp4.mp4'
    output_path = '/content/output.mp4'
    model_path = '/content/roop/models/inswapper_128.onnx'

    if not os.path.exists(source_path):
        print(f"Source image not found: {source_path}")
        return
    if not os.path.exists(target_path):
        print(f"Target video not found: {target_path}")
        return
    if not os.path.exists(model_path):
        print(f"ONNX Model not found at {model_path}")
        return

    print("Initializing FaceAnalysis with CPU providers to bypass broken downloads...")
    app = FaceAnalysis(name='buffalo_l', providers=['CPUExecutionProvider'])
    app.prepare(ctx_id=0, det_size=(640, 640))

    print("Loading face swapper model...")
    swapper = insightface.model_zoo.get_model(model_path, download=False, download_zip=False)

    # Detect face from source image
    source_img = cv2.imread(source_path)
    source_faces = app.get(source_img)
    if not source_faces:
        print("No face detected in the source image!")
        return
    source_face = source_faces[0]
    print("Source face detected successfully!")

    # Open target video
    cap = cv2.VideoCapture(target_path)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

    print(f"Processing video ({total_frames} frames)...")
    frame_count = 0
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        # Detect faces in target frame
        target_faces = app.get(frame)
        if target_faces:
            for face in target_faces:
                frame = swapper.get(frame, face, source_face, paste_back=True)

        out.write(frame)
        frame_count += 1
        if frame_count % 10 == 0:
            print(f"Processed {frame_count}/{total_frames} frames...")

    cap.release()
    out.release()
    print(f"Face swap frames processing complete. Output video generated.")

    # Automatically merge original audio track back
    merge_audio(output_path, target_path)
    print(f"All tasks complete! Saved to: {output_path}")

if __name__ == '__main__':
    main()
````

````
import subprocess
import os

def merge_audio():
    original_video = '/content/contentvideo_40.mp4.mp4'
    swapped_video = '/content/output.mp4'
    final_output = '/content/output_with_audio.mp4'

    if not os.path.exists(original_video):
        print("Original video not found!")
        return
    if not os.path.exists(swapped_video):
        print("Swapped video not found!")
        return

    print("Merging original audio into the face-swapped video...")

    # ffmpeg command to take video from swapped_video, audio from original_video, and copy codecs without re-encoding
    cmd = [
        'ffmpeg', '-y',
        '-i', swapped_video,
        '-i', original_video,
        '-c:v', 'copy',
        '-c:a', 'aac',
        '-map', '0:v:0',
        '-map', '1:a:0',
        '-shortest',
        final_output
    ]

    result = subprocess.run(cmd, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)

    if result.returncode == 0:
        print(f"Success! Your video with audio is saved at: {final_output}")
        # Replace the original silent output with the one containing audio
        os.replace(final_output, swapped_video)
        print("Successfully updated /content/output.mp4 with audio.")
    else:
        print("Error occurred while merging audio:")
        print(result.stderr)

merge_audio()
````

````
from IPython.display import HTML
from base64 import b64encode
import os

def play_video(video_path):
    if not os.path.exists(video_path):
        print(f"Video file not found at: {video_path}")
        return

    print("Loading video into Colab player...")
    mp4 = open(video_path, 'rb').read()
    data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
    return HTML(f"""
    <video width=600 controls autoplay>
          <source src="{data_url}" type="video/mp4">
    </video>
    """)

play_video('/content/output.mp4')
````
